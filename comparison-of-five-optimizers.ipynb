# Comparison of Five Optimizers for Logistic Regression - SMS Spam Classification

This notebook compares five optimization algorithms used to train Logistic Regression models for SMS Spam Classification. The comparison is based on their training loss, test loss, accuracy, precision, recall, and F1 score.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Performance Comparison Table

This section presents a comparison of the five optimization algorithms based on their training loss, test loss, accuracy, precision, recall, and F1 score. This comparison helps identify the optimizer that achieves the best classification performance.

```python
results = {
    'Optimizer': [
        'GD with Momentum',
        'SGD with Momentum',
        'Mini-batch GD with Momentum',
        'RMSProp with Momentum',
        'Adam with Momentum'
    ],

    'Training Loss': [
        0.160429,
        0.013986,
        0.133682,
        0.002979,
        0.145184
    ],

    'Test Loss': [
        0.163319,
        0.058189,
        0.138414,
        0.061451,
        0.169555
    ],

    'Accuracy': [
        0.949776,
        0.982960,
        0.960538,
        0.984753,
        0.970404
    ],

    'Precision': [
        0.989474,
        0.985075,
        0.990654,
        0.978261,
        0.975410
    ],

    'Recall': [
        0.630872,
        0.885906,
        0.711409,
        0.906040,
        0.798658
    ],

    'F1 Score': [
        0.770492,
        0.932862,
        0.828125,
        0.940767,
        0.878229
    ]
}

comparison_df = pd.DataFrame(results)

comparison_df
```

## Accuracy Comparison

This bar chart compares the accuracy of the five optimization algorithms. It shows how accurately each model classifies SMS messages as Ham or Spam.

```python
plt.figure(figsize=(10, 5))

plt.bar(comparison_df['Optimizer'], comparison_df['Accuracy'])

plt.title('Accuracy Comparison of Five Optimizers')
plt.xlabel('Optimizer')
plt.ylabel('Accuracy')
plt.ylim(0, 1)
plt.xticks(rotation=20, ha='right')

for i, value in enumerate(comparison_df['Accuracy']):
    plt.text(i, value + 0.01, f'{value:.3f}', ha='center')

plt.tight_layout()
plt.show()
```

# Precision, Recall and F1 Score Comparison

This chart compares Precision, Recall, and F1 Score across the five optimization algorithms to evaluate their ability to identify Spam messages correctly.

```python
metrics = ['Precision', 'Recall', 'F1 Score']

comparison_df.set_index('Optimizer')[metrics].plot(
    kind='bar',
    figsize=(12, 6),
    width=0.8
)

plt.title('Precision, Recall and F1 Score Comparison')
plt.xlabel('Optimizer')
plt.ylabel('Score')
plt.ylim(0, 1)
plt.xticks(rotation=20, ha='right')
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

# Training Loss Comparison

This chart compares the final training loss values of the five optimization algorithms. Lower training loss indicates that the model achieved a better fit to the training data.

```python
plt.figure(figsize=(10, 5))

plt.bar(comparison_df['Optimizer'], comparison_df['Training Loss'])

plt.title('Training Loss Comparison of Five Optimizers')
plt.xlabel('Optimizer')
plt.ylabel('Training Loss')
plt.xticks(rotation=20, ha='right')

for i, value in enumerate(comparison_df['Training Loss']):
    plt.text(i, value + 0.003, f'{value:.4f}', ha='center')

plt.tight_layout()
plt.show()
```

## Test Loss Comparison

This chart compares the Binary Cross-Entropy Loss of the five models on the test dataset. Lower test loss indicates better prediction performance on unseen data.

```python
plt.figure(figsize=(10, 5))

plt.bar(comparison_df['Optimizer'], comparison_df['Test Loss'])

plt.title('Test Loss Comparison of Five Optimizers')
plt.xlabel('Optimizer')
plt.ylabel('Test Loss')
plt.xticks(rotation=20, ha='right')

for i, value in enumerate(comparison_df['Test Loss']):
    plt.text(i, value + 0.003, f'{value:.4f}', ha='center')

plt.tight_layout()
plt.show()
```

# Conclusion

The comparison shows that each optimization algorithm achieved different results when training the Logistic Regression model for SMS Spam Classification.

RMSProp with Momentum achieved the highest Accuracy, Recall, and F1 Score, as well as the lowest Training Loss. SGD with Momentum achieved the lowest Test Loss, showing strong performance on unseen data.

Overall, RMSProp with Momentum provided the best balance among the evaluation metrics in this experiment, while SGD with Momentum also demonstrated competitive performance.
